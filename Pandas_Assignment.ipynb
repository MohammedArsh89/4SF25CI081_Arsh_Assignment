# Pandas Assignment 

```python
import pandas as pd
import numpy as np
np.random.seed(42)
```

## 1. DataFrame with product column

```python
df1 = pd.DataFrame(np.random.randint(1, 10, (5, 3)), columns=['A', 'B', 'C'])
df1['Product'] = df1['A'] * df1['B']
print(df1)
```

## 2. Row-wise and column-wise sums

```python
df2 = pd.DataFrame(np.random.randint(1, 10, (4, 3)), columns=['A', 'B', 'C'])
print(df2)
print("Row-wise sum:\\n", df2.sum(axis=1))
print("Column-wise sum:\\n", df2.sum(axis=0))
```

## 3. Fill NaN with respective column mean

```python
df3 = pd.DataFrame(np.random.randint(1, 10, (5, 3)), columns=['A', 'B', 'C'], dtype=float)
df3.iloc[0, 1] = np.nan
df3.iloc[2, 2] = np.nan
df3.iloc[4, 0] = np.nan
print("Before:\\n", df3)
df3 = df3.fillna(df3.mean())
print("After mean imputation:\\n", df3)
```

## 4. Drop rows containing any NaN

```python
df4 = pd.DataFrame(np.random.randint(1, 20, (6, 4)), columns=['A', 'B', 'C', 'D'], dtype=float)
df4.iloc[1, 2] = np.nan
df4.iloc[4, 0] = np.nan
print("Before:\\n", df4)
print("After dropna:\\n", df4.dropna())
```

## 5. Group by Category and calculate sum and mean

```python
df5 = pd.DataFrame({'Category': np.random.choice(['A', 'B', 'C'], 10),
                    'Value': np.random.randint(10, 100, 10)})
print(df5)
print(df5.groupby('Category')['Value'].agg(['sum', 'mean']))
```

## 6. Total sales for each category

```python
df6 = pd.DataFrame({'Product': np.random.choice(['Pen', 'Book', 'Pencil'], 10),
                    'Category': np.random.choice(['A', 'B', 'C'], 10),
                    'Sales': np.random.randint(100, 1000, 10)})
print(df6)
print(df6.groupby('Category')['Sales'].sum())
```

## 7. Merge two DataFrames

```python
df7a = pd.DataFrame({'ID': [1, 2, 3], 'Name': ['Ram', 'Sam', 'Raj']})
df7b = pd.DataFrame({'ID': [1, 2, 3], 'Marks': [90, 85, 95]})
print(pd.merge(df7a, df7b, on='ID'))
```

## 8. MultiIndex indexing and slicing

```python
idx = pd.MultiIndex.from_tuples([('A','One'), ('A','Two'), ('B','One'), ('B','Two')],
                               names=['Category', 'SubCategory'])
df8 = pd.DataFrame({'Value': [10, 20, 30, 40]}, index=idx)
print(df8)
print("Category A:\\n", df8.loc['A'])
print("A, One:\\n", df8.loc[('A', 'One')])
print("Slice A to B:\\n", df8.loc['A':'B'])
```

## 9. MultiIndex aggregation

```python
df9 = pd.DataFrame({'Category': np.random.choice(['A', 'B', 'C'], 12),
                    'SubCategory': np.random.choice(['X', 'Y'], 12),
                    'Value': np.random.randint(10, 100, 12)})
df9 = df9.set_index(['Category', 'SubCategory'])
print(df9)
print(df9.groupby(level=['Category', 'SubCategory'])['Value'].sum())
```

## 10. Double all DataFrame values

```python
df10 = pd.DataFrame(np.random.randint(1, 10, (5, 3)), columns=['A', 'B', 'C'])
print("Original:\\n", df10)
print("Doubled:\\n", df10.apply(lambda col: col * 2))
```

## 11. Add row-wise sum using lambda

```python
df11 = pd.DataFrame(np.random.randint(1, 10, (6, 3)), columns=['A', 'B', 'C'])
df11['Sum'] = df11.apply(lambda row: row['A'] + row['B'] + row['C'], axis=1)
print(df11)
```

## 12. Convert strings to uppercase

```python
s12 = pd.Series(['apple', 'banana', 'mango', 'orange', 'grapes'])
print(s12.str.upper())
```

## 13. Extract first three characters

```python
s13 = pd.Series(['apple', 'banana', 'mango', 'orange', 'grapes'])
print(s13.str[:3])
```
